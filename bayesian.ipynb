```python
# Bayesian Network Example
# Problem: Disease Diagnosis using Bayesian Network

# Install:
# pip install pgmpy

from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination

# ---------------- MODEL CREATION ----------------

# Create Bayesian Network Structure
# Fever -> Flu
# Cough -> Flu

model = DiscreteBayesianNetwork([
    ('Flu', 'Fever'),
    ('Flu', 'Cough')
])

# ---------------- DEFINE PROBABILITIES ----------------

# Probability of Flu
cpd_flu = TabularCPD(
    variable='Flu',
    variable_card=2,
    values=[[0.95], [0.05]]
)

# Probability of Fever given Flu
cpd_fever = TabularCPD(
    variable='Fever',
    variable_card=2,
    values=[
        [0.90, 0.20],   # No Fever
        [0.10, 0.80]    # Fever
    ],
    evidence=['Flu'],
    evidence_card=[2]
)

# Probability of Cough given Flu
cpd_cough = TabularCPD(
    variable='Cough',
    variable_card=2,
    values=[
        [0.85, 0.30],   # No Cough
        [0.15, 0.70]    # Cough
    ],
    evidence=['Flu'],
    evidence_card=[2]
)

# ---------------- ADD CPDs ----------------

model.add_cpds(cpd_flu, cpd_fever, cpd_cough)

# ---------------- VALIDATE MODEL ----------------

print("Model is valid:", model.check_model())

# ---------------- INFERENCE ----------------

inference = VariableElimination(model)

# Query:
# Probability of Flu given Fever and Cough

result = inference.query(
    variables=['Flu'],
    evidence={
        'Fever': 1,
        'Cough': 1
    }
)

# ---------------- OUTPUT ----------------

print("\nProbability of Flu given Fever and Cough:\n")
print(result)
```

```python
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination

# ---------------- MODEL ----------------

model = DiscreteBayesianNetwork([
    ('Flu', 'Fever'),
    ('Flu', 'Cough')
])

# ---------------- CPDs ----------------

cpd_flu = TabularCPD(
    variable='Flu',
    variable_card=2,
    values=[[0.95], [0.05]]
)

cpd_fever = TabularCPD(
    variable='Fever',
    variable_card=2,
    values=[
        [0.90, 0.20],
        [0.10, 0.80]
    ],
    evidence=['Flu'],
    evidence_card=[2]
)

cpd_cough = TabularCPD(
    variable='Cough',
    variable_card=2,
    values=[
        [0.85, 0.30],
        [0.15, 0.70]
    ],
    evidence=['Flu'],
    evidence_card=[2]
)

# ---------------- ADD CPDs ----------------

model.add_cpds(
    cpd_flu,
    cpd_fever,
    cpd_cough
)

# ---------------- CHECK MODEL ----------------

print("Model Valid:", model.check_model())

# ---------------- INFERENCE ----------------

inference = VariableElimination(model)

result = inference.query(
    variables=['Flu'],
    evidence={
        'Fever': 1,
        'Cough': 1
    }
)

# ---------------- OUTPUT ----------------

print(result)
```
