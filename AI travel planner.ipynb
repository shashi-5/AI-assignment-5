```python
# AI Based Travel Planner using Knowledge Base
# Domain Used: Tourist Places + Food Recommendation
# Simple Rule-Based AI Planner

# ---------------- KNOWLEDGE BASE ----------------

tourist_places = {
    "Hyderabad": {
        "places": [
            "Charminar",
            "Golconda Fort",
            "Ramoji Film City",
            "Hussain Sagar"
        ],
        "food": [
            "Hyderabadi Biryani",
            "Haleem",
            "Double Ka Meetha"
        ],
        "budget_per_day": 3000
    },

    "Goa": {
        "places": [
            "Baga Beach",
            "Dudhsagar Falls",
            "Fort Aguada",
            "Anjuna Beach"
        ],
        "food": [
            "Goan Fish Curry",
            "Prawn Balchao",
            "Bebinca"
        ],
        "budget_per_day": 5000
    },

    "Delhi": {
        "places": [
            "India Gate",
            "Red Fort",
            "Qutub Minar",
            "Lotus Temple"
        ],
        "food": [
            "Chole Bhature",
            "Butter Chicken",
            "Paratha"
        ],
        "budget_per_day": 4000
    }
}

# ---------------- AI TRAVEL PLANNER ----------------

def generate_plan(city, days, interest):

    if city not in tourist_places:
        return "City not available in knowledge base."

    data = tourist_places[city]

    print("\n========== AI TRAVEL PLAN ==========")
    print("Destination:", city)
    print("Duration:", days, "days")
    print("Interest:", interest)

    # Cost Assessment
    total_cost = data["budget_per_day"] * days

    print("\nEstimated Budget: ₹", total_cost)

    # Personalized Tour Plan
    print("\nRecommended Tourist Places:")

    places = data["places"]

    for i in range(min(days, len(places))):
        print(f"Day {i+1}: Visit {places[i]}")

    # Food Recommendation
    print("\nRecommended Local Foods:")

    for food in data["food"]:
        print("-", food)

    print("\n====================================")


# ---------------- TEST CASES ----------------

# Test Case 1
generate_plan(
    city="Hyderabad",
    days=3,
    interest="Historical Places"
)

# Test Case 2
generate_plan(
    city="Goa",
    days=2,
    interest="Beaches"
)

# Test Case 3
generate_plan(
    city="Delhi",
    days=4,
    interest="Culture and Food"
)
```
