```python
# Heuristic Alpha-Beta Search
# Depth-limited search with heuristic evaluation

import math

PLAYER = 'X'
OPPONENT = 'O'
EMPTY = '_'

def heuristic(board):
    score = 0

    # Simple heuristic:
    # +1 for each PLAYER mark
    # -1 for each OPPONENT mark

    for row in board:
        for cell in row:
            if cell == PLAYER:
                score += 1
            elif cell == OPPONENT:
                score -= 1

    return score

def is_moves_left(board):
    for row in board:
        if EMPTY in row:
            return True
    return False

def heuristic_alphabeta(board, depth, alpha, beta, is_max, max_depth):

    if depth == max_depth or not is_moves_left(board):
        return heuristic(board)

    if is_max:
        best = -math.inf

        for i in range(3):
            for j in range(3):
                if board[i][j] == EMPTY:
                    board[i][j] = PLAYER

                    best = max(
                        best,
                        heuristic_alphabeta(
                            board, depth + 1,
                            alpha, beta,
                            False, max_depth
                        )
                    )

                    board[i][j] = EMPTY

                    alpha = max(alpha, best)

                    if beta <= alpha:
                        break

        return best

    else:
        best = math.inf

        for i in range(3):
            for j in range(3):
                if board[i][j] == EMPTY:
                    board[i][j] = OPPONENT

                    best = min(
                        best,
                        heuristic_alphabeta(
                            board, depth + 1,
                            alpha, beta,
                            True, max_depth
                        )
                    )

                    board[i][j] = EMPTY

                    beta = min(beta, best)

                    if beta <= alpha:
                        break

        return best

def find_best_move(board, max_depth):
    best_val = -math.inf
    best_move = (-1, -1)

    for i in range(3):
        for j in range(3):
            if board[i][j] == EMPTY:
                board[i][j] = PLAYER

                move_val = heuristic_alphabeta(
                    board, 0,
                    -math.inf, math.inf,
                    False, max_depth
                )

                board[i][j] = EMPTY

                if move_val > best_val:
                    best_move = (i, j)
                    best_val = move_val

    return best_move

# ---------------- TEST CASE 1 ----------------

board = [
    ['X', 'O', '_'],
    ['_', 'O', '_'],
    ['_', '_', 'X']
]

best_move = find_best_move(board, 3)

print("Best Move using Heuristic Alpha-Beta:", best_move)
```
